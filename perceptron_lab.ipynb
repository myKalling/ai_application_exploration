Build Your First Perceptron with Python
A Perceptron is a computer neuron that mimics human neurons. 

```python
# Define the activation function (step function)
x1 = 1
x2 = 0

# Weights for the inputs
w1 = 0.8
w2 = -0.4

# Bias term
bias = -0.2
```

```python
# Define the activation function (step function)
def step_activation(value):
    if value >= 0:
        return 1
    else:
        return 0
```

```python
# Calculate the weighted sum (z)
z = (x1 * w1) + (x2 * w2) + bias
print("Weighted sum (z):", z)

output = step_activation(z)
print("Perceptron output:", output)
```

```python
def perceptron(x1, x2, w1, w2, bias):
    # weighted sum for the defined Perceptron (z)
    weighted_sum = (x1 * w1) + (x2 * w2) + bias
    step_output = step_activation(weighted_sum)
    return weighted_sum, step_output
```

```python
weighted_sum, prediction = perceptron(1, 0, 0.8, -0.4, -0.2)
print("Weighted sum (z):", weighted_sum)
print("Perceptron output:", prediction)
```

```python
# For Loop Randomized with imported random module

import random

def step_activation(z):
    return 1 if z >= 0 else 0

def perceptron(x1, x2, w1, w2, bias):
    # Weighted sum for the defined Perceptron (z)
    weighted_sum = (x1 * w1) + (x2 * w2) + bias
    step_output = step_activation(weighted_sum)
    return weighted_sum, step_output

# For loop to randomize the perceptron
for i in range(10):
    w1 = random.uniform(-1, 1)
    w2 = random.uniform(-1, 1)
    bias = random.uniform(-1, 1)

    weighted_sum, prediction = perceptron(1, 0, w1, w2, bias)

    print(f"Run {i + 1}:")
    print(f"  Weight 1 (w1): {w1:.2f}")
    print(f"  Weight 2 (w2): {w2:.2f}")
    print(f"  Bias: {bias:.2f}")
    print(f"  Weighted sum (z): {weighted_sum:.2f}")
    print(f"  Perceptron output: {prediction}")
    print("-" * 30)
```

1. What does a positive weight do to the weighted sum?
A positive weight increases the weighted sum when its corresponding input is positive. This means the input contributes positively toward activating the perceptron.

2. What does a negative weight do to the weighted sum?
A negative weight decreases the weighted sum when its corresponding input is positive. This means the input contributes negatively and can make activation less likely.

3. What is the purpose of the bias?
The purpose of the bias is to shift the decision boundary of the perceptron. It allows the model to produce an output even when all inputs are zero and helps control when the activation threshold is reached.

4. What does the step activation function do?
The step activation function converts the weighted sum into a binary output. It returns 1 when the weighted sum is greater than or equal to zero and returns 0 when the weighted sum is less than zero.

5. Which input combinations produced an output of 1?
The input combinations that produced an output of 1 were those whose weighted sum was greater than or equal to zero. Based on the perceptron results, these combinations were (1,0) and (1,1).

6. What type of classification problem produces two possible classes?
A binary classification problem produces two possible classes. Examples include Yes/No, True/False, Pass/Fail, or Spam/Not Spam.
