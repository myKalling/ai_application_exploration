Build Your First Perceptron with Python
A Perceptron is a computer neuron that mimics human neurons. 

```python
# Define the activation function (step function)
x1 = 1
x2 = 0

# Weights for the inputs
w1 = 0.8
w2 = -0.4

# Bias term
bias = -0.2
```

```python
# Define the activation function (step function)
def step_activation(value):
    if value >= 0:
        return 1
    else:
        return 0
```

```python
# Calculate the weighted sum (z)
z = (x1 * w1) + (x2 * w2) + bias
print("Weighted sum (z):", z)

output = step_activation(z)
print("Perceptron output:", output)
```

```python
def perceptron(x1, x2, w1, w2, bias):
    # weighted sum for the defined Perceptron (z)
    weighted_sum = (x1 * w1) + (x2 * w2) + bias
    step_output = step_activation(weighted_sum)
    return weighted_sum, step_output
```

```python
weighted_sum, prediction = perceptron(1, 0, 0.8, -0.4, -0.2)
print("Weighted sum (z):", weighted_sum)
print("Perceptron output:", prediction)
```

```python
# For Loop Randomized with imported random module

import random

def step_activation(z):
    return 1 if z >= 0 else 0

def perceptron(x1, x2, w1, w2, bias):
    # Weighted sum for the defined Perceptron (z)
    weighted_sum = (x1 * w1) + (x2 * w2) + bias
    step_output = step_activation(weighted_sum)
    return weighted_sum, step_output

# For loop to randomize the perceptron
for i in range(10):
    w1 = random.uniform(-1, 1)
    w2 = random.uniform(-1, 1)
    bias = random.uniform(-1, 1)

    weighted_sum, prediction = perceptron(1, 0, w1, w2, bias)

    print(f"Run {i + 1}:")
    print(f"  Weight 1 (w1): {w1:.2f}")
    print(f"  Weight 2 (w2): {w2:.2f}")
    print(f"  Bias: {bias:.2f}")
    print(f"  Weighted sum (z): {weighted_sum:.2f}")
    print(f"  Perceptron output: {prediction}")
    print("-" * 30)
```
